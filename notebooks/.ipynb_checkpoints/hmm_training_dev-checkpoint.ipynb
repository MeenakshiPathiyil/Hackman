```python
# Load corpus.txt
with open('D:\Hackman\data\corpus.txt', 'r', encoding='utf-8') as f:
    corpus = [line.strip() for line in f if line.strip()]

# Check few samples and total length
print("Sample words:", corpus[:10])
print("Total words:", len(corpus))
```

```python
from collections import defaultdict

grouped_train = defaultdict(list)
for w in corpus:
    grouped_train[len(w)].append(w)

for length, group in list(grouped_train.items())[:24]:
    print(f"Length {length}: {len(group)} words")
#prints count of first 5 word-length groups
```

```python
from hmmlearn.hmm import CategoricalHMM
import numpy as np

# Build letter vocabulary with start/end tokens
letters = list("abcdefghijklmnopqrstuvwxyz")
tokens = ['<S>', '<E>']
vocab = tokens + letters
char_to_int = {c: i for i, c in enumerate(vocab)}
int_to_char = {i: c for c, i in char_to_int.items()}
```

```python
# Encode each word as integer sequence with start/end
sequences = []
for word in corpus:
    seq = [char_to_int['<S>']] + [char_to_int[ch] for ch in word if ch in letters] + [char_to_int['<E>']]
    sequences.append(seq)
```

```python
# Combine into numpy format
lengths = [len(seq) for seq in sequences]
X = np.concatenate([np.array(seq) for seq in sequences]).reshape(-1, 1)

print("Total sequences:", len(sequences))
print("Total observations:", len(X))
print("Sample encoded sequence:", sequences[0][:10])
```

```python
# Train Categorical HMM
n_states = 30  # adjustable (20–40 recommended)
model = CategoricalHMM(n_components=n_states, n_iter=30, random_state=42, verbose=True)
model.fit(X, lengths)

print("HMM training complete!")
print("Number of states:", model.n_components)
```

```python
import joblib

# Save the model
joblib.dump({
    "model": model,
    "char_to_int": char_to_int,
    "int_to_char": int_to_char
}, "../models/hmm_model.pkl")

print("HMM model and mappings saved successfully!")

print("HMM model saved successfully!")
```

```python
import numpy as np

def get_letter_probabilities(model, pattern, guessed_letters, char_to_int, int_to_char):
    letters = [c for c in char_to_int if c not in ['<S>', '<E>']]
    letter_probs = {}

    for letter in letters:
        if letter in guessed_letters:
            continue
        
        # Replace blanks with the candidate letter
        test_word = pattern.replace('_', letter)
        seq = [char_to_int['<S>']] + [char_to_int[ch] for ch in test_word if ch in char_to_int] + [char_to_int['<E>']]
        X = np.array(seq).reshape(-1, 1)
        
        try:
            logprob = model.score(X)
            letter_probs[letter] = np.exp(logprob)  # Convert log-likelihood to normal prob
        except:
            letter_probs[letter] = 0.0  # Unlikely / invalid sequence

    # Normalize
    total = sum(letter_probs.values())
    if total > 0:
        for k in letter_probs:
            letter_probs[k] /= total
    
    return dict(sorted(letter_probs.items(), key=lambda x: x[1], reverse=True))

# Example usage
pattern = "_pp_e"
guessed_letters = set("ape")
probs = get_letter_probabilities(model, pattern, guessed_letters, char_to_int, int_to_char)
print("Top predictions:")
for l, p in list(probs.items())[:10]:
    print(f"{l}: {p:.4f}")
```

```python
import joblib
import numpy as np
import random
from collections import defaultdict
from hmmlearn.hmm import CategoricalHMM

# Load the trained model and mappings
model = joblib.load("../models/hmm_model.pkl")
# char_to_int = joblib.load("models/char_to_int.pkl")
# int_to_char = {v: k for k, v in char_to_int.items()}

# Load test corpus
with open("data/test.txt", "r", encoding="utf-8") as f:
    test_words = [w.strip().lower() for w in f if w.strip()]

letters = list("abcdefghijklmnopqrstuvwxyz")

def predict_missing_letter(word_with_mask):
    """Return top predicted letter for '_' in word_with_mask."""
    # Identify missing letter position
    pos = word_with_mask.index("_")
    results = {}

    for letter in letters:
        test_word = word_with_mask.replace("_", letter)
        seq = [char_to_int['<S>']] + [char_to_int[ch] for ch in test_word] + [char_to_int['<E>']]
        X_test = np.array(seq).reshape(-1, 1)
        results[letter] = model.score(X_test)

    # Return the best predicted letter
    return max(results, key=results.get)

# Evaluate
total, correct = 0, 0
for word in test_words:
    if len(word) <= 2:
        continue  # skip too short
    # Randomly mask one letter (not first/last)
    pos = random.randint(1, len(word) - 2)
    masked = word[:pos] + "_" + word[pos + 1:]

    predicted = predict_missing_letter(masked)
    true = word[pos]

    if predicted == true:
        correct += 1
    total += 1

success_rate = correct / total if total > 0 else 0
print(f"✅ Success rate on test set: {success_rate * 100:.2f}% ({correct}/{total})")
```
