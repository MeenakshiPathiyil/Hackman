```python
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]

with open(filename, 'r', encoding='utf-8') as f:
    words = f.read().splitlines()

print("Sample before cleaning:", words[:10])
print("Total words before cleaning:", len(words))
```

```python
import re

cleaned = []
seen = set()

for w in words:
    w = w.strip().lower()
    if re.fullmatch(r'[a-z]+', w) and w not in seen:
        cleaned.append(w)
        seen.add(w)

print("Sample after cleaning:", cleaned[:10])
print("Total words after cleaning:", len(cleaned))
```

```python
print("Unique words before cleaning:", len(set(words)))
print("Duplicates removed:", len(words) - len(set(words)))
```

```python
original_set = set([w.strip().lower() for w in words if w.strip()])
cleaned_set = set(cleaned)

dropped = original_set - cleaned_set

print("Total dropped words:", len(dropped))
print("Sample dropped words:", list(dropped)[:20])
```

```python
import re

# Load corpus
with open("corpus.txt", "r") as f:
    words = [w.strip() for w in f if w.strip()]

# Remove words with spaces and keep only alphabetic
filtered = [w.lower() for w in words if re.match("^[a-zA-Z]+$", w)]

# Keep one copy per word (deduplicate)
seen = set()
final_words = []
for w in filtered:
    if w not in seen:
        seen.add(w)
        final_words.append(w)

# Save cleaned corpus
with open("cleaned_corpus.txt", "w") as f:
    for w in final_words:
        f.write(w + "\n")

print("Total words before cleaning:", len(words))
print("Total words after cleaning:", len(final_words))
print("Sample after cleaning:", final_words[:10])
```
